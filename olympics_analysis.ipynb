```python
import pandas as pd
import sqlite3
import os

# Build absolute paths to bypass folder/file name confusion
base_dir = r"D:\SportsStats"
athlete_path = os.path.join(base_dir, "athlete_events.csv")
noc_path = os.path.join(base_dir, "noc_regions.csv")

try:
    # 1. Load the unzipped CSV data into DataFrames
    athlete_df = pd.read_csv(athlete_path)
    noc_df = pd.read_csv(noc_path)
    print("✅ CSV files successfully loaded into DataFrames!")
    
    # 2. Open local SQLite connection
    conn = sqlite3.connect(os.path.join(base_dir, 'olympics.db'))

    # 3. Store DataFrames as queryable database tables
    athlete_df.to_sql('athlete_events', conn, if_exists='replace', index=False)
    noc_df.to_sql('noc_regions', conn, if_exists='replace', index=False)
    print("💾 Both tables successfully written to 'olympics.db'!")
    
    conn.close()

except PermissionError:
    print("❌ Still blocked! Please close Excel or any background app viewing the CSV file.")
except FileNotFoundError as e:
    print(f"❌ File not found error: {e}")
```

```python
import pandas as pd
import sqlite3

# 1. Connect back to your freshly made database file
conn = sqlite3.connect(r"D:\SportsStats\olympics.db")

# 2. Write your SQL query inside the triple quotes
# This query joins both tables to find the top 5 countries with the most Olympic medals
query = """
SELECT 
    r.region AS country, 
    COUNT(a.medal) AS total_medals
FROM athlete_events a
LEFT JOIN noc_regions r ON a.noc = r.noc
WHERE a.medal IS NOT NULL
GROUP BY r.region
ORDER BY total_medals DESC
LIMIT 5;
"""

# 3. Use Pandas to run the SQL query and format it as a beautiful table
df_result = pd.read_sql_query(query, conn)

# 4. Display the table view output
display(df_result)

# Close the connection line when finished
conn.close()
```
